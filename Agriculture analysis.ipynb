```python
def convert_to_billions(value):

  if pd.isna(value):
    return None

  value = str(value).replace('$', '').strip()

  if value.endswith('B'):
    return float(value[:-1])

  elif value.endswith('M'):
    return float(value[:-1]) / 1000

  else:
    return float(value)
```

### Handling Categorical Object Columns with One-Hot Encoding

The remaining object columns (`State`, `District`, `Crop`, `Season`, `Irrigation_Method`) are categorical. To convert them into a numerical format suitable for analysis or modeling, I will apply one-hot encoding. This process creates new binary (0 or 1) columns for each unique category in the original column. After encoding, the original object columns will be dropped.

```python
categorical_cols = df.select_dtypes(include='object').columns.tolist()

# Exclude 'Farm_ID' as it's an identifier and not a categorical feature for one-hot encoding.
# A numerical 'Farm_Id' has already been created from it.
if 'Farm_ID' in categorical_cols:
    categorical_cols.remove('Farm_ID')

# Perform one-hot encoding on the identified categorical columns
df_processed = pd.get_dummies(df, columns=categorical_cols, drop_first=False, dtype=float)

# Drop the original 'Farm_ID' column now that 'Farm_Id' (numerical) is available and categorical are encoded
df_processed = df_processed.drop('Farm_ID', axis=1)

display(df_processed.head())
display(df_processed.info())

# Impute missing values for numerical columns using the median
for col in ['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']:
    if col in df_processed.columns:
        median_val = df_processed[col].median()
        df_processed[col].fillna(median_val, inplace=True)

display(df_processed.isnull().sum())
```

```python
#OUTLIER INVESTIGATION
outlier_columns = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Soil_pH',
    'Fertilizer_kg_ha',
    'Yield_Tonnes_Ha',
    'Profit_INR',
    'Water_Used_m3',
    'Disease_Pest_Risk_pct'
]

outlier_summary = []

for col in outlier_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        'Variable': col,
        'Lower_Bound': lower,
        'Upper_Bound': upper,
        'Outlier_Count': count,
        'Outlier_Percentage': count / len(df) * 100
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df.sort_values(
    'Outlier_Count',
    ascending=False
)
```
